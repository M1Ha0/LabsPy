```python
import numpy as np
x = np.arange(0, 1.01, 0.01)
y = np.round(np.sin(x)**2 - np.cos(2 * x), 2)
res=y[y > 0]
print(res)
```

Задание 3.1

1)

```python
arr1 = np.array(res, dtype=int)
print(arr1)
```

2)

```python
arr2 = np.linspace(1.0, 10.0, len(arr1), dtype=float)
print(arr2)
```

3)

```python
arr3 = np.random.gumbel(loc=0.0, scale=1.0, size=len(arr1)).astype(float)
print(arr3)
```

5)

```python
op1 = ((arr1 + arr2 + arr3) * 5).astype(float)
print("1(Сложить)")
print(op1)
op2 = (arr1 - arr2).astype(float)
print("2(Вычесть)")
print(op2)
op3 = (arr1 * arr3).astype(float)
print("3(Умножить)")
print(op3)
op4 = (arr1 / arr2).astype(float)
print("4(Разделить)")
print(op4)
op5 = (arr1 ** 3).astype(float)
print("5(Возвести в степень)")
print(op5)
```
