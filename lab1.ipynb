```python
arr = [
    [2,-3,0],
    [-1,0,-5],
    [0,4,0],
    [5,-2,1],
]

for i in range(len(arr[0])):
    fn = None
    lz = None
    for j in range(len(arr)):
        if arr[j][i] < 0:
            fn = j
            break
    for j in range(len(arr) -1, -1, -1):
        if arr[j][i] == 0:
            lz = j
            break
    if fn is not None and lz is not None:
        arr[fn][i], arr[lz][i] =arr[lz][i], arr[fn][i]

for i in arr:
    print(i)
```

Задание 4

```python

students_languages = [
    {"Russian", "English", "Japanese"},
    {"Russian", "English", "German"},
    {"Russian", "Spanish"},
]

common_languages = set(students_languages[0])
all_languages = set()

for student in students_languages:
    common_languages &= student
    all_languages |= student


sorted_common = sorted(common_languages)
sorted_all = sorted(all_languages)


print(len(sorted_common))
for lang in sorted_common:
    print(lang)

print(len(sorted_all))
for lang in sorted_all:
    print(lang)
```

Задание 5

```python
from collections import defaultdict

english_to_latin = {
    "apple": ["malum", "pomum"],
    "fruit": ["baca", "bacca", "popum"],
    "punishment": ["malum"],
}

latin_to_english = defaultdict(list)

for en_word, lat_words in english_to_latin.items():
    for lat_word in lat_words:
        latin_to_english[lat_word].append(en_word)

print(len(latin_to_english))

for lat_word in sorted(latin_to_english.keys()):
    sorted_en_words = sorted(latin_to_english[lat_word])
    translations = ", ".join(sorted_en_words)
    print(f"{lat_word} - {translations}")
```
