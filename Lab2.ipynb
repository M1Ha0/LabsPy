```python
class Employee:
   def __init__(self, surname, salary, year_of_birth):
       self.surname = surname #строка
       self.salary = salary #число
       self.year_of_birth = year_of_birth #число

   def print_age(self):
       print('Возраст работника:', 2026-self.year_of_birth , 'лет')
   def print_day(self):
        print('Лет до исполнеия сотруднику 50 лет:', 50-(2026-self.year_of_birth))


emploer1 = Employee("Sergio", 20000, 2000)
emploer1.print_age()
emploer1.print_day()
```

```python
class EmployeeCompany(Employee):
   def __init__(self, surname, salary, year_of_birth, post):
       super().__init__(surname,salary,year_of_birth)
       self.post = post

   def print_salary_developer(self):
       if self.post=="Программист":
           self.salary*=1.2
       return print("Оклад работника",self.salary)
emploerCompany3 = EmployeeCompany("nekitich", 20000, 2000, "Программист")
emploerCompany3.print_salary_developer()
```

```python
class Partiy:
   def __init__(self, name, size, percent):
       self.name = name #строка
       self.size = size #число
       self.percent = percent #число

   def print_Q(self):
       print("Q=",0.3*self.size+0.7*self.percent)
class Partiy2level(Partiy):
    def __init__(self, name, size, percent, lastsize):
        super().__init__(name,size,percent)
        self.lastsize = lastsize
    def print_Q2level(self):
        if self.size>self.lastsize:
            return print("Q=", 1.2*(0.3*self.size+0.7*self.percent))
        else:
            return print("Q=", 0.8*(0.3*self.size+0.7*self.percent))
partiy1 = Partiy("Y", 12, 11)
partiy1.print_Q()
partiy2 = Partiy2level("Y", 13, 11, 14)
partiy2.print_Q2level()
```
